## Series와 DataFrame

- **Series**: 1차원. 엑셀의 **열 하나**
- **DataFrame**: 2차원 표. **Series 여러 개를 옆으로 붙인 것**

df["Name"]  -> Series  (열 하나를 꺼냄)  
df[["Name"]] -> DataFrame (열 목록을 꺼냄, 결과가 표)

- pd.Series(데이터, index, name): 데이터를 시리즈로 변환
- pd.DataFrame(데이터, index, columns): 데이터를 데이터프레임으로 변환

```python
data = {
    'Name':['Jackson', 'Emma', 'Noah', 'James'],
    'Weight':[68, 74, 77, 78],
    'Option':[True, True, False, False],
    'Rate':[0.21, 1.1, 0.89, 0.91],
    'Type':['A','A','C','C']
}
```

| Name | Weight | Option | Rate | Type |
|---|---|---|---|---|
| Jackson | 68 | True | 0.21 | A |
| Emma | 74 | True | 1.10 | A |
| Noah | 77 | False | 0.89 | C |
| James | 78 | False | 0.91 | C |


### 데이터프레임으로 변환
- pd.DataFrame(데이터)

```python
df = pd.DataFrame(data)
df
```

```python
print("df의 타입: ", type(df))
```

### DataFrame -> Numpy 배열

- 데이터프레임.to_numpy()  

컬럼 이름과 인덱스가 사라지고 값만 남는다

```python
df_converted = df.to_numpy()
df_converted
```

```python
print("df_converted.dtype: ", df_converted.dtype)
```

#### 2. info()

```python
df.info()
```

### info() 읽는 법

| 항목 | 의미 |
|---|---|
| RangeIndex: 891 entries | 전체 891행 |
| Non-Null Count | 값이 있는 행의 개수 |
| Dtype | 자료형 (object = 문자열) |

**전체 891과 다른 컬럼이 결측치가 있는 컬럼이다.**
- Age: 714 -> 177개 비어 있음
- Cabin: 204 -> 687개 비어 있음 (77%)
- Embarked: 889 -> 2개 비어 있음

#### 3. describe()

```python
df.describe()
```

### describe() 읽는 법

- 숫자 컬럼만 나온다 (Name, Sex 같은 문자 컬럼은 제외)
- 25% / 50% / 75% = **사분위수**. 값을 작은 순으로 줄 세웠을 때 1/4, 2/4, 3/4 지점
  - Age 25% = 20.1 -> 승객의 1/4이 20세 이하
  - 50%는 중앙값(median)

### 여기서 읽어낸 것

1. **count가 컬럼마다 다르다** — Age만 714. 결측치가 여기서도 드러난다
2. **Fare는 평균 32.20, 중앙값 14.45, 최대 512.33**
   → 평균이 중앙값의 2배. 소수의 비싼 티켓이 평균을 끌어올렸다
   → 이런 컬럼은 "평균 요금 32달러"라고 말하면 실제와 동떨어진다
3. **Survived의 평균 0.3838 = 생존율 38.4%**
   0/1 값만 있는 컬럼은 평균이 곧 비율이다
