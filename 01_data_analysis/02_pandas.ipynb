## Series와 DataFrame

- **Series**: 1차원. 엑셀의 **열 하나**
- **DataFrame**: 2차원 표. **Series 여러 개를 옆으로 붙인 것**

df["Name"]  -> Series  (열 하나를 꺼냄)  
df[["Name"]] -> DataFrame (열 목록을 꺼냄, 결과가 표)

- pd.Series(데이터, index, name): 데이터를 시리즈로 변환
- pd.DataFrame(데이터, index, columns): 데이터를 데이터프레임으로 변환

```python
data = {
    'Name':['Jackson', 'Emma', 'Noah', 'James'],
    'Weight':[68, 74, 77, 78],
    'Option':[True, True, False, False],
    'Rate':[0.21, 1.1, 0.89, 0.91],
    'Type':['A','A','C','C']
}
```

| Emma | 74 | True | 1.10 | A |

## loc / iloc

| | 기준 | 예시 |
|---|---|---|
| loc | **레이블(이름)** | df.loc[0, "Age"] |
| iloc | **위치(정수 순서)** | df.iloc[0, 5] |

타이타닉은 인덱스가 0,1,2...라 둘이 같아 보이지만,
**행을 걸러내는 순간 달라진다.**

슬라이싱도 다르다
- df.loc[0:3]  -> 0,1,2,3  (끝 포함)
- df.iloc[0:3] -> 0,1,2    (끝 제외, 파이썬 기본과 동일)

## 조건 필터링

- `and`가 아니라 `&`를 쓴다
  df["Age"]>50 의 결과는 True/False가 891개 든 Series다.
  `and`는 값 하나끼리만 비교 가능 -> 에러.
  `&`는 같은 자리끼리 비교한다 (Numpy 논리 인덱싱과 같은 원리)

- **괄호 필수**
  `&`가 `>`보다 먼저 계산되기 때문
  df["Age"] > 50 & df["Age"] < 60      ❌
  (df["Age"] > 50) & (df["Age"] < 60)  ⭕

- or는 `|`, 부정은 `~`

```python
print(df.loc[0:3].shape)    # (4, ...) 끝 포함
print(df.iloc[0:3].shape)   # (3, ...) 끝 제외
```

```python
df.sort_values("Fare", ascending=False).head()
```

```python
# Age 컬럼의 값이 50초과 60미만인 행만 뽑아서 보기
tmp_df = df.loc[(df["Age"]>50) & (df["Age"]<60)]
print(tmp_df.shape)
tmp_df.head()
```

```python
# 걸러낸 뒤에는 인덱스가 띄엄띄엄해진다
print(tmp_df.index[:5])

print(tmp_df.iloc[0]["Name"])   # 위에서 첫 번째 행
# print(tmp_df.loc[0])          # KeyError — 0번은 걸러져서 없다
```

## 그룹화 (groupby)

원리: 1) 쪼개기 → 2) 적용 → 3) 결합

df.groupby("Sex")["Age"].mean() 이 하는 일

1. **쪼개기**: 891행을 Sex 값에 따라 male 577행 / female 314행으로 나눈다
2. **적용**: 각 덩어리에서 Age의 평균을 따로 구한다
3. **결합**: 두 결과를 하나의 Series로 합친다

`df.groupby(기준컬럼)[대상컬럼].집계함수()`

```python
# 성별에 따른 'Age'의 평균 구하기
df.groupby("Sex")['Age'].mean()
```

```python
# 성별마다의 'Pclass'별 'Age'의 평균 구하기
df.groupby(["Sex", "Pclass"])["Age"].mean()
```

```python
# 여러 통계를 한 번에
df.groupby("Sex")["Age"].agg(["mean", "median", "count"])
```

**count를 같이 보는 이유**

Sex별 인원은 male 577 / female 314인데
count는 453 / 261로 나온다. **결측치를 빼고 센 것**이다.
평균이 몇 명 기준인지 항상 같이 확인해야 한다.

```python
# groupby 결과는 인덱스가 2단으로 쌓인 형태
# reset_index()로 평범한 표로 되돌린다
df.groupby(["Sex", "Pclass"])["Age"].mean().reset_index()
```

```python
# 전체 컬럼 결측치 한 번에
print(df.isna().sum())

# 비율로 보기
print(df.isna().mean().sort_values(ascending=False))
```

```python
# 확인: fillna는 원본을 바꾸지 않는다
print("처리 전 평균 :", df["Age"].mean())
_ = df.fillna(0)
print("fillna 후 평균:", df["Age"].mean())   # 그대로다
```

```python
# 0으로 일괄 처리하면 어떻게 되는지 확인
print("원본 평균      :", df["Age"].mean())
print("0으로 채운 평균:", df["Age"].fillna(0).mean())
```

```python
# 컬럼 성격에 맞게 처리한다

# Age: 숫자 -> 중앙값
# 평균이 아니라 중앙값을 쓰는 이유: 극단값에 덜 흔들린다
df["Age"] = df["Age"].fillna(df["Age"].median())

# Embarked: 범주형 -> 최빈값
# mode()는 동점일 수 있어 Series를 반환하므로 [0]으로 첫 값을 꺼낸다
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Cabin: 687/891 = 77% 결측 -> 채울 근거가 없으니 컬럼 삭제
df = df.drop(columns=["Cabin"])

print(df.isna().sum())
```

### 결측치 처리 기준

| 상황 | 처리 |
|---|---|
| 숫자 컬럼 | 중앙값(median). 평균은 극단값에 흔들린다 |
| 범주형 컬럼 | 최빈값(mode) |
| 결측이 너무 많음(대략 50% 이상) | 컬럼 삭제 |
| 행이 몇 개 안 됨 | dropna()로 행 삭제 |

**inplace=True는 쓰지 않는다.** pandas 3.x에서 동작이 바뀌었고,
`df = df.어쩌구()` 재대입이 어디서 뭐가 바뀌었는지 코드에 드러난다.

```python
# 나이를 구간으로 묶기 (파생 변수)
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 12, 20, 40, 60, 100],
    labels=["아동", "청소년", "청년", "중년", "노년"]
)

# 나이대별 생존율
df.groupby("AgeGroup", observed=True)["Survived"].agg(["mean", "count"])
```

### 주의: 이 결과의 한계

Age 결측치 177개를 중앙값 28로 채웠다.
28은 (20, 40] 구간이므로 **177명이 전부 "청년"에 들어갔다.**
청년 562명 중 실제 나이를 아는 사람은 385명뿐이다.

즉 청년의 생존율 0.365는 만들어낸 값이 섞인 숫자다.
결측치를 채우는 처리는 이렇게 **뒤쪽 분석 결과까지 오염시킨다.**

대안: 결측치를 채우지 않고 pd.cut에 그대로 넣으면
NaN은 어느 구간에도 안 들어가서 "모름"으로 분리된다.
